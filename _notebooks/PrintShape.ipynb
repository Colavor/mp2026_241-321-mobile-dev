```
// (5)
/// <summary>
/// Нарисовать прямоугольник в консоль
/// </summary>
/// <param name="width">Ширина фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>
string[] GetRectangle
(
  int width,
  int height,
  char symbol = '*',
  bool isFilled = true,
  char symbolClear = ' '
)
{
  List<string> result = new();
  for (int r = 0; r < height; r++)
  {
    if (isFilled || r == 0 || r == height - 1)
      Console.WriteLine(new string(symbol, width));
    else
      Console.WriteLine(symbol + new string(symbolClear, width - 2) + symbol);
  }
  return result.ToArray();
}

var shape = GetRectangle(25, 4);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать пирамиду
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>

string[] GetPyramid(
    int height,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        int spaces = height - r - 1;
        int symbols = r * 2 + 1;

        if (isFilled)
            result.Add(
                new string(symbolClear, spaces) +
                new string(symbol, symbols));
        else
            result.Add(
                new string(symbolClear, spaces) +
                symbol +
                (r == 0 ? "" : new string(symbolClear, symbols - 2) + symbol));
    }

    return result.ToArray();
}

var shape = GetPyramid(6);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать ромб
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>

string[] GetDiamond(
    int height,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        int middle = height / 2;
        int width = r <= middle ? r * 2 + 1 : (height - r - 1) * 2 + 1;
        int spaces = (height - width) / 2;

        if (isFilled)
            result.Add(
                new string(symbolClear, spaces) +
                new string(symbol, width));
        else
            result.Add(
                new string(symbolClear, spaces) +
                symbol +
                (width > 1 ? new string(symbolClear, width - 2) + symbol : ""));
    }

    return result.ToArray();
}

var shape = GetDiamond(7);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать песочные часы
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>

string[] GetHourglass(
    int height,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        int width = r < height / 2
            ? height - r * 2
            : (r - height / 2) * 2 + 1;

        int spaces = (height - width) / 2;

        if (isFilled)
            result.Add(
                new string(symbolClear, spaces) +
                new string(symbol, width));
        else
            result.Add(
                new string(symbolClear, spaces) +
                symbol +
                (width > 1 ? new string(symbolClear, width - 2) + symbol : ""));
    }

    return result.ToArray();
}

var shape = GetHourglass(7);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать зебру
/// </summary>
/// <param name="width">Ширина фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ полосы</param>
/// <param name="symbolClear">Символ очистки</param>
/// <param name="stripeWidth">Толщина полосы</param>

string[] GetZebra(
    int width,
    int height,
    char symbol = '*',
    char symbolClear = ' ',
    int stripeWidth = 1)
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        char current = r / stripeWidth % 2 == 0
            ? symbol
            : symbolClear;

        result.Add(new string(current, width));
    }

    return result.ToArray();
}

var shape = GetZebra(20, 8, '#', '.', 2);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```
