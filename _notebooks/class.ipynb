```
// (1)

class Person
{
    // Поля
    public string name;
    public int age = 18; // По умолчанию

    // Методы
    public string Say() => "Привет";
}

var p = new Person();
p.name = "юра";
p.age = 20;
Console.WriteLine($"{p.name} ({p.age})");

// через инициализатор объекта
var p2 = new Person { name = "Петя", age = 19 };
Console.WriteLine($"{p2.name} ({p2.age})");
```

```
// (3) Свойства

class Person
{
    // Поля
    private string name;
    private int age;

    // Свойства
    public string Name
    {
        get
        {
            return name;
        }
        set
        {
            name = value;
        }
    }

    public int Age { get => age; set => age = value; }
}

var p = new Person();
p.Name = "юра";
p.Age = 20;
Console.WriteLine($"{p.Name} ({p.Age})");
```

```
// (4) Автоматические свойства

class Person
{
    // Поля
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }

    //
    public string Nickname { get => $"{Name} {Age}"; }
}

var p = new Person();
p.Name = "Юра";
p.Age = 20;
Console.WriteLine($"{p.Name} ({p.Age})");
Console.WriteLine(p.Nickname);
```

```
// (5) Обычный конструктор

class Person
{
    // Поля
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }

    // Конструктор
    public Person(string name, int age)
    {
        Name = name;
        Age = age;
    }
}

var p = new Person("Ваня", 17);
p.Name = "Юра";
p.Age = 20;
Console.WriteLine($"{p.Name} ({p.Age})");
```

```
// (7) Первичный конструктор + Метод ToString

class Person(string name, int age)
{
    public string Name { get; set; } = name;
    public int Age { get; set; } = age;

    public override string ToString() => $"Person: {Name} ({Age})";
}

var p = new Person("Ваня", 20);
Console.WriteLine(p);
```

```
// (9) Модификатор required (начиная с C# 11.0)

// required - нужно (обязательно) установить:
// - либо в конструкторе,
// - либо в инициализаторе.

class Person
{
    public required string Name { get; set; }
    public required int Age { get; set; }
}

// var p = new Person(); // нельзя
var p = new Person() { Name = "Петя", Age = 18 };
p.Name = "Тузик";
```

```
// (10) Деконструктор (не путать с деструктором)

class Person
{
    public string Name { get; set; }
    public int Age { get; set; }

    public void Deconstruct(out string name, out int age)
    {
        name = Name;
        age = Age;
    }

    public void Deconstruct(out string name, out int age, out string text)
    {
        name = Name;
        age = Age;
        text = "Привет";
    }
}

var p = new Person() { Name = "Юра", Age = 19 };

(string name, int age) = p;
Console.WriteLine(name);
Console.WriteLine(age);

// (string name1, int age1, string text) = p;
(_, _, string text) = p;
Console.WriteLine(text);
```

```
// (13) Наследование + первичный конструктор

class Person(string name)
{
    public string Name { get; set; } = name;
}

class Employee(string name, string company) : Person(name)
{
    public string Company { get; set; }

    public Employee(string name, string company) : base(name)
    {
        Company = company;
    }
}

// var e = new Employee("Ваня", "Яндекс");
var e = new Employee(name: "Ваня", company: "Яндекс");
```

```
// (18) Индексаторы

class Bag(string[] items)
{
    private string[] items = items;

    // Индексатор
    public string this[Index index]
    {
        get => items[index];
        set => items[index] = value;
    }
}

var bag = new Bag(["Ноутбук", "Наушники", "Термос"]);

Console.WriteLine(bag[0]);
Console.WriteLine(bag[^1]);

bag[^1] = "Шапка";
Console.WriteLine(bag[^1]);
```

```
class Code(string text)
{
    private string text = text;

    public int Length { get => text.Length; }

    // Индексатор
    public string this[int index]
    {
        get => $"{text[index]}{text[^(index + 1)]}";
    }
}

var code = new Code("Hello");

Console.WriteLine(code[0]);    // Ho
Console.WriteLine(code[1]);    // el
// Console.WriteLine(code[-1]);

for (int i = 0; i < code.Length; i++)
    Console.WriteLine($"code[{i}] = {code[i]}");
```
